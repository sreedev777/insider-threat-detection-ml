```python
import pandas as pd

logon = pd.read_csv("../data/logon.csv")
file_activity = pd.read_csv("../data/file.csv")
device = pd.read_csv("../data/device.csv")

print("Logon:", logon.shape)
print("File:", file_activity.shape)
print("Device:", device.shape)
```

```python
print("LOGON COLUMNS:")
print(logon.columns.tolist())

print("\nFILE COLUMNS:")
print(file_activity.columns.tolist())

print("\nDEVICE COLUMNS:")
print(device.columns.tolist())
```

```python
print("LOGON DATA TYPES:")
print(logon.dtypes)

print("\nFILE DATA TYPES:")
print(file_activity.dtypes)

print("\nDEVICE DATA TYPES:")
print(device.dtypes)
```

```python
print("LOGON MISSING VALUES:")
print(logon.isnull().sum())

print("\nFILE MISSING VALUES:")
print(file_activity.isnull().sum())

print("\nDEVICE MISSING VALUES:")
print(device.isnull().sum())
```

```python
logon["date"] = pd.to_datetime(logon["date"])
file_activity["date"] = pd.to_datetime(file_activity["date"])
device["date"] = pd.to_datetime(device["date"])

print("LOGON DATE RANGE:")
print(logon["date"].min(), "to", logon["date"].max())

print("\nFILE DATE RANGE:")
print(file_activity["date"].min(), "to", file_activity["date"].max())

print("\nDEVICE DATE RANGE:")
print(device["date"].min(), "to", device["date"].max())
```

```python
print("Unique users in logon:", logon["user"].nunique())
print("Unique users in file:", file_activity["user"].nunique())
print("Unique users in device:", device["user"].nunique())
```

- **Scenario ID/name**: Scenario 2
  - **Malicious user ID**: CCH0959
  - **Start date**: 08/02/2010 10:34:31
  - **End date**: 09/30/2010 15:04:03
  - **Malicious activity**: User begins surfing job websites and soliciting employment from a competitor. Before leaving the company, they use a thumb drive (at markedly higher rates than their previous activity) to steal data.

- **Scenario ID/name**: Scenario 3
  - **Malicious user ID**: KTW0365
  - **Start date**: 07/08/2010 11:51:00
  - **End date**: 07/09/2010 20:54:16
  - **Malicious activity**: System administrator becomes disgruntled. Downloads a keylogger and uses a thumb drive to transfer it to his supervisor's machine. The next day, he uses the collected keylogs to log in as his supervisor and send out an alarming mass email, causing panic in the organization. He leaves the organization immediately.
